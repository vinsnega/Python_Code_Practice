```python
#A grading system determines whether a student has passed or failed based on their score.
mark=int(input("Enter your marks:"))
if mark>=40:
    print("Student has passed")
else:
    print("Student has not passed")
```

```python
#A retail store offers a 20% discount if a customer’s total order exceeds $100. Write logic to calculate the final amount to be paid after applying the discount.
bill=int(input("Enter customer bill amount:$"))
if bill>100:
    discount=(bill*20)/100
    bill=bill-discount
    print("your bill amount is :$" ,bill)
else:
    print("No discount")
```

```python
#A banking system processes withdrawal requests and ensures the user has enough balance.
#Write logic to check if a user has enough balance before allowing a withdrawal and update the remaining balance accordingly
Savings =200000
withdrawal_amt=int(input("Enter amount:"))
if withdrawal_amt<Savings:
    print("Withdrawal request approved")
    print("Remaining balance amount is :",(Savings-withdrawal_amt))
else:
    print("No Sufficient balance")
```

```python
#A calendar system verifies whether a given year is a leap year based on standard leap year rules.
year=int(input("Enter a year :"))
if year%4==0 and year%100!=0:
    print(f"{year} is leap year")
elif(year%400==0):
    print(f"{year} is leap year")
elif(year%100==0 and year%400!=0):
    print(f"{year} is not a leap year")
else:
    print(f"{year} is not a leap year")
```
